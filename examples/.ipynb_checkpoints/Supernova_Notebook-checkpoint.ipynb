### Importing necessary libraries

```python
import numpy as np
import matplotlib.pyplot as plt
from spextractor import Spextractor
import json
import csv
```

### Making .txt file from the .json file

```python
json_file = '/home/raditraian1107/Supernova_Project/examples/ASASSN-14hr.json'
final_output = '/home/raditraian1107/Supernova_Project/examples/ASASSN-14hr.dat'

print(f"Reading JSON from: {json_file}")

with open(json_file, 'r') as f:
    data_structure = json.load(f)

raw_data = data_structure['ASASSN-14hr']['spectra'][0]['data']

data_array = np.array(raw_data, dtype=float)

wavelength = data_array[:, 0]
flux = data_array[:, 1]
error = data_array[:, 2]

mask = np.isclose(flux, error)

if np.any(mask):
    print(f"Found {np.sum(mask)} lines with identical Flux/Error. Setting Error to 0.")
    error[mask] = 0  # Set bad errors to 0
else:
    print("Your file is good to go!")

clean_data = np.column_stack((wavelength, flux, error))
np.savetxt(final_output, clean_data, fmt='%s')

print("The file is ready.")
print(f"NEW FILENAME TO USE: {final_output}")
```

### Extracting Silicon Layer Velocity and pEW by Spextractor

```python
fn = '/home/raditraian1107/Supernova_Project/examples/ASASSN-14hr.dat'

with open(json_file, 'r') as f:
    meta_data = json.load(f)
z_str = meta_data['ASASSN-14hr']['redshift'][0]['value']
z = float(z_str)
print(f"Found Redshift: z = {z}")

spex = Spextractor(fn, z=z, plot=True)
spex.create_model(model_type='SKI')

features = ('Si II 6150A', 'Si II 5800A')
spex.process(features)

print(f"{'Feature':<15} | {'Velocity (km/s)':<25} | {'pEW (Angstroms)':<25}")
print("-" * 70)

for feat in features:
    v = spex.vel[feat]
    v_err = spex.vel_err[feat]
    p = spex.pew[feat]
    p_err = spex.pew_err[feat]
    
    print(f"{feat:<15} | {v:.3f} +/- {v_err:.3f}          | {p:.3f} +/- {p_err:.3f}")

fig, ax = spex.plot

plt.tight_layout()
fig.savefig('Ia_example.png', dpi=300)
plt.show()
```
